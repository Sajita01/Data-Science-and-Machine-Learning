```python
# Series: It is a one-dimensional array-like object that can store any data type such as integers, strings, floats, or even Python objects. I
```

```python
# pandas.Series(data=None, index=None, dtype=None, name=None, copy=False)
import pandas as pd
a = ['g','e','e','k','s']
res = pd.Series(a)
print(res)
```

```python
# Series holding the Int data type
import pandas as pd
data = [1, 2, 3, 4, 5]
res = pd.Series(data)
print(res)
```

```python
# Series holding the dictionary
import pandas as pd
a = { 'Id': 1013,'Name': 'Mohan','State': 'ktm','Age': 24}
res = pd.Series(a)
print(res)
```

```python
# Dataframe: combination of data and column is data frame
# Syntax: pandas.DataFrame(data=None, index=None, columns=None, dtype=None, copy=False)
```

```python
# Creating a dataFrame from a list
import pandas as pd
a = ['Python','Pandas','Numpy']
df = pd.DataFrame(a, columns=['Tech'])
print(df)
```

```python
# Creating a dataFrame from a dictionary
a = {
 'Name': ['Tom', 'Nick', 'Krish', 'Jack'],
 'Age': [20, 21, 19, 18]
}
res = pd.DataFrame(a)
print(res)
```

```python
# Selecting columns and rows in a dataFrame
import pandas as pd
a = {
'Name': ['Jai','Princi','Gaurav','Anuj'],
'Age': [27, 24, 22, 32],
'Address': ['Delhi','Kanpur','Allahabad','Kannauj'],
'Qualification': ['Msc','MA','MCA','Phd']}
df = pd.DataFrame(a)
print(df[['Name','Qualification']])
```
