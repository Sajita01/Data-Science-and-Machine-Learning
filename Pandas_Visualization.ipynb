```python
import pandas as pd
```

```python
df = pd.read_csv("/content/drive/MyDrive/dsml/dataset.csv")
df
```

```python
import matplotlib.pyplot as plt
```

```python
df["Sex"].value_counts().plot.bar()
plt.title("Gender Distrubution")
plt.show()
```

```python
df['Survived'].value_counts().plot(
    kind='bar',
    title='Survival Distribution',
    figsize=(6, 4),
    yticks=range(0, 600, 100),
    xlabel='Survived',
    ylabel='Count',
    legend=False
)

plt.show()
```

```python
df.groupby('Sex')['Survived'].value_counts().unstack().plot(
    kind='bar',
    title='Count of survived passenger',
    figsize=(6, 4),
    xlabel='survive(0=No,1=yes)',
    ylabel='No. of passenger',
    yticks=range(0, 600, 100),
    xticks=(0, 1),
    color=['grey', 'green'],
    legend=False
)

plt.legend(['Male', 'Female'])
plt.show()
```

```python
pd.crosstab(df['Sex'], df['Survived']).plot
```

```python
pd.crosstab(df['Sex'], df['Survived']).plot(
    kind='bar',
    title='Survival Count by Gender',
    figsize=(6, 4),
    xlabel='Gender',
    ylabel='Number of Passengers',
    color=['grey', 'green'],
    yticks=range(0, 501, 100),
    rot=0
)

# Add legend
plt.legend(['Not Survived', 'Survived'])

# Display the chart
plt.show()
```

```python
df['Embarked'].value_counts().plot(
    kind='pie',
    title='Passenger Distribution by Embarkation Port',
    figsize=(4, 4),
    colors=['skyblue', 'lightgreen', 'orange'],
    labels=['S', 'C', 'Q'],
    ylabel='',
    legend=True
)
plt.show()
```

```python
df['Embarked'].value_counts().plot(
    kind='pie',
    title='Passenger Distribution by Embarkation Port',
    figsize=(4, 4),
    colors=['skyblue', 'lightgreen', 'orange'],
    labels=['S', 'C', 'Q'],
    ylabel='',
    legend=True,
    autopct='%1.1f%%',
)
plt.show()
```

```python
# Male and Female survival distribution

df.groupby('Sex')['Survived'].value_counts().unstack().plot(
    kind='pie',
    subplots=True,
    title=['Female Survival Distribution', 'Male Survival Distribution'],
    figsize=(10, 5),
    autopct='%1.1f%%',
    colors=['grey', 'green'],
    legend=True,
    ylabel=''
)

plt.show()
```

```python
# line graph
pd.crosstab(df['Sex'], df['Survived']).plot(
    kind='line',
    title='Survival Distribution by Gender',
    figsize=(7, 5),
    xlabel='Gender',
    ylabel='Number of Passengers',
    marker='o',
    color=['grey', 'green']
)

plt.legend(['Not Survived', 'Survived'])
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

df.plot(
    kind='scatter',
    x='Age',
    y='Fare',
    title='Relationship between Age and Fare',
    figsize=(7, 5),
    color='blue',
    alpha=0.6
)

plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid()
plt.show()
```

```python
# Histogram: it gives frequency
df['Age'].plot(
    kind='hist',
    title='Age Distribution of Titanic Passengers',
    figsize=(7, 5),
    bins=20,
    color='skyblue',
    edgecolor='black'
)

plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
df["Age"].plot.hist(bins=20,grid=True)
```

```python
#  box plot
df['Age'].plot(
    kind='box',
    title='Age Distribution of Titanic Passengers',
    figsize=(6, 5),
    color='blue'
)

plt.ylabel('Age')
plt.show()
```
